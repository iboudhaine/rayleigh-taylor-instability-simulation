# CUDA run

Build the CUDA runner, simulate the Rayleigh-Taylor instability, and animate the result. Designed to run on Kaggle with a GPU accelerator enabled.

```python
!git clone https://github.com/iboudhaine/rayleigh-taylor-instability-simulation.git
%cd rayleigh-taylor-instability-simulation
```

```python
!nvidia-smi
!nvcc --version
```

```python
!make -C src/cuda
```

```python
!./src/cuda/run \
  --Nx 1024 --Ny 2048 \
  --t-end 3.0 --scheme rk2 \
  --save-interval 800 \
  --output-dir results/cuda_vis_1024x2048_t3 \
  --block-x 16 --block-y 16 \
  --seed 42 --overwrite
```

```python
!python scripts/animate.py results/cuda_vis_1024x2048_t3 --out /kaggle/working/rt.mp4
```
